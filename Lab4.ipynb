```python
import numpy as np
import statistics as mode
```

```python
data=[45, 67, 89, 45, 67, 45, 89, 90, 67, 45]
print("Dataset: ", data)
```

```python
mean = np.mean(data)
median = np.median(data)
mode_value = mode.mode(data)

print("Mean: ", mean)
print("Median: ", median)
print("Mode: ", mode_value)

data_range=np.max(data)-np.min(data)
variance=np.var(data)
std_dev=np.std(data)

print("Range: ", data_range)
print("Variance: ", variance)
print("Standard Deviation: ", std_dev)
```

```python
import pandas as pd

data={
    'Age':[25, 36, 45, 56, 45],
    'Study_Hours':[2, 7, 6, 5, 6],
    'Attendance':[70, 95, 60, 90, 78],
    'Sleep_Hours':[6, 8, 8, 9, 10],
    'Marks':[70, 75, 80, 85, 90] #target variable
}

df=pd.DataFrame(data)
```

```python
correlation=df.corr()['Marks']
print(correlation)
```

```python
threshold=0.8
selected_features=correlation[abs(correlation)>threshold].index.tolist()
```

```python
selected_features.remove('Marks') #remove target variable from selected features
print("Selected Features: ", selected_features)
```
