```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
data={
    "Age": [20, 25, 30,45],
    "Salary": [2000, 3000, 5000, 4000]
}
```

```python
print(data)
df=pd.DataFrame(data)
```

```python
print(df)
```

<h1>Standardization</h1>

```python
scaler=StandardScaler()
df_standardized=scaler.fit_transform(df)
df_standardized=pd.DataFrame(df_standardized,columns=df.columns)
print(df_standardized)
```

<h1>Min-Max Scaling</h1>

```python
from sklearn.preprocessing import MinMaxScaler
min_max_scaler=MinMaxScaler()
df_min_max_scaled=min_max_scaler.fit_transform(df)
df_min_max_scaled=pd.DataFrame(df_min_max_scaled,columns=df.columns)
print(df_min_max_scaled)
```

<h1>SMOTE-(Synthetic Minority Over-sampling Technique)</h1>

```python
from collections import Counter
from sklearn.datasets import make_classification
from imblearn.over_sampling import SMOTE
```

```python
X,y=make_classification(
    n_samples=1000,
    n_features=5,
    n_informative=3,
    n_redundant=0,
    weights=[0.90,0.10],
    random_state=42
)

```

```python
print("Before SMOTE:",Counter(y))
```

```python
smote=SMOTE(random_state=42)
X_resampled,y_resampled=smote.fit_resample(X,y)
```

```python
print("After SMOTE:",Counter(y_resampled))
```
