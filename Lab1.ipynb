```python
#Priyanka Swain
import matplotlib.pyplot as plt        
from sklearn.datasets import load_breast_cancer

data=load_breast_cancer()


plt.figure(figsize=(7,3))
plt.hist(data.data[:,0],bins=30)
plt.hist(data.data[:,0],bins=30)
plt.xlabel('Mean Radius')
plt.ylabel('Frequency')
plt.show()
```

```python
#Priyanka Swain
plt.figure(figsize=(7,3))
plt.hist(data.data[data.target==0,0],bins=30,label='Malignant',alpha=0.5)
plt.hist(data.data[data.target==1,0],bins=30,label='Benign',alpha=0.5)
plt.xlabel('Mean Radius')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
#Priyanka Swain
import matplotlib.pyplot as plt 
import seaborn as sns       
from sklearn.datasets import load_breast_cancer
data=load_breast_cancer()

plt.figure(figsize=(7,3))
sns.kdeplot(data.data[:,0],shade=True)
plt.xlabel('Mean Radius')
plt.ylabel('Density')
plt.show()
```

```python
#Priyanka Swain
plt.figure(figsize=(7,3))
sns.kdeplot(data.data[data.target==0,0],label='Malignant',shade=True)
sns.kdeplot(data.data[data.target==1,0],label='Benign',shade=True)
plt.xlabel('Mean Radius')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
#Priyanka Swain
import matplotlib.pyplot as plt 
import seaborn as sns       
from sklearn.datasets import load_iris
iris=load_iris()
data=iris.data
target=iris.target
plt.figure(figsize=(7,3))
sns.boxenplot(x=target,y=data[:,0])
plt.xlabel("Species")
plt.ylabel("Sepal Length")
plt.show()
```
