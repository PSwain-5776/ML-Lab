```python
import pandas as pd
```

```python
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Priya", "Neha", "Sonu"],
    "Age": [21, 22, None, 24, 22, None, 24],
    "City": ["Delhi", "Mumbai", "Pune", "Mumbai", "Delhi", "Noida", "Kolkata"]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
df = df.drop(columns=["City"])
print(df)
```

```python
df = df.dropna()
print(df)
```

```python
df = df.drop_duplicates()
print(df)
```
